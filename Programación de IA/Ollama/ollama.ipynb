```python
import requests

url = "http://localhost:11434/api/chat"

datos = {
    "model": "qwen3:4b",
    #"model": "qwen3:4b-instruct",
    "messages": [
        {
            "role": "user",
            "content": "Cuánto es 1 más 2"
        }
    ],
    "think": False,
    "stream": False,
    "keep_alive": "10m",
    "options": {
        "num_predict": 150
    }
}

r = requests.post(url, json=datos, timeout=120)
r.raise_for_status()

resultado = r.json()

print("Respuesta:", resultado["message"]["content"])
print("Carga:", round(resultado["load_duration"] / 1e9, 2), "s")
print("Total:", round(resultado["total_duration"] / 1e9, 2), "s")

if resultado["eval_duration"] > 0:
    velocidad = resultado["eval_count"] / (
        resultado["eval_duration"] / 1e9
    )
    print("Velocidad:", round(velocidad, 2), "tokens/s")
```
