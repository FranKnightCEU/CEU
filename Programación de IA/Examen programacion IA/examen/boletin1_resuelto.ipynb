## Ejercicio 1

Desarrolla un programa que solicite una contraseña y compruebe si cumple los siguientes requisitos:
- Tener al menos 8 caracteres.
- Contener al menos una letra mayúscula.
- Contener al menos una letra minúscula.
- Contener al menos un número.
- Contener al menos un carácter especial.

El programa debe indicar si la contraseña es válida y, en caso contrario, mostrar qué requisitos incumple (pueden ser varios de los mencionados simultáneamente).

NOTA: isupper(), islower(), isdigit(), isalnum() son métodos que se pueden aplicar sobre cadenas de caracteres para validar.

---

**Explicación breve:** Guardamos cada requisito incumplido en una lista y al final mostramos todos los errores juntos.

````python
# Pedimos una contraseña y guardamos los requisitos que no cumple.
contrasena = input("Introduce una contraseña: ")
errores = []

if len(contrasena) < 8:
    errores.append("Debe tener al menos 8 caracteres")
if not any(caracter.isupper() for caracter in contrasena):
    errores.append("Debe contener una mayúscula")
if not any(caracter.islower() for caracter in contrasena):
    errores.append("Debe contener una minúscula")
if not any(caracter.isdigit() for caracter in contrasena):
    errores.append("Debe contener un número")
if not any(not caracter.isalnum() for caracter in contrasena):
    errores.append("Debe contener un carácter especial")

if len(errores) == 0:
    print("Contraseña válida")
else:
    print("Contraseña no válida:")
    for error in errores:
        print("-", error)
````

## Ejercicio 2

Un servidor registra el tiempo de respuesta de sus últimas peticiones HTTP, expresado en milisegundos.

Desarrolla un programa que, a partir de una lista de tiempos, muestre:
- Tiempo medio de respuesta.
- Tiempo máximo y mínimo.
- Número de peticiones que superan los 500 ms. Vamos a considerar que entonces son rápidas y las que sean igual o menor son lentas.
- Porcentaje de peticiones lentas.
- Si el tiempo medio supera los 300 ms, mostrar una alerta de rendimiento.

Usa por ejemplo estos valores de tiempos: 120, 340, 560, 230, 890, 120, 450, 670, 210 y 180

---

**Explicación breve:** Calculamos estadísticas con `sum`, `len`, `max` y `min`; mantenemos la clasificación rápida/lenta tal como aparece en el enunciado.

````python
tiempos = [120, 340, 560, 230, 890, 120, 450, 670, 210, 180]
media = sum(tiempos) / len(tiempos)
# El enunciado considera rápidas las superiores a 500 ms.
rapidas = sum(1 for tiempo in tiempos if tiempo > 500)
lentas = len(tiempos) - rapidas

print(f"Tiempo medio: {media:.2f} ms")
print("Máximo:", max(tiempos), "ms")
print("Mínimo:", min(tiempos), "ms")
print("Peticiones superiores a 500 ms:", rapidas)
print(f"Peticiones lentas: {lentas / len(tiempos) * 100:.2f}%")
if media > 300:
    print("Alerta de rendimiento: media superior a 300 ms")
````

## Ejercicio 3

Desarrolla un programa que reciba un texto y muestre las palabras que aparecen más de una vez (TODAS ELLAS), indicando cuántas veces se repite cada una.

La comparación debe ignorar mayúsculas y minúsculas.

**Ejemplo**

- Entrada: Python es un lenguaje muy potente, además Python tiene una sintaxis muy sencilla
- Salida: 
    ```python
    python: 2
    muy: 2
    ```

---

**Explicación breve:** Normalizamos el texto a minúsculas, contamos cada palabra en un diccionario y mostramos solo las repetidas.

````python
texto = input("Escribe un texto: ")
# Pasamos a minúsculas y quitamos puntuación básica.
for signo in ",.;:¡!¿?()":
    texto = texto.replace(signo, " ")
palabras = texto.lower().split()
conteos = {}
for palabra in palabras:
    conteos[palabra] = conteos.get(palabra, 0) + 1

print("Palabras repetidas:")
for palabra, cantidad in conteos.items():
    if cantidad > 1:
        print(f"{palabra}: {cantidad}")
````

## Ejercicio 4

Una tienda dispone de un inventario con el nombre de cada producto y su cantidad disponible.

Desarrolla un programa que permita:
1. Consultar las existencias de un producto.
2. Añadir unidades a un producto.
3. Retirar unidades de un producto.
4. Mostrar todos los productos.
5. Salir del programa.

No se deben permitir cantidades negativas ni retirar más unidades de las disponibles.

Usa por ejemplo los siguientes datos de inventario:
- teclado: 10,
- ratón: 15,
- monitor: 5

---

**Explicación breve:** Un diccionario guarda las existencias y un menú `while` permite consultar y modificarlas con validaciones.

````python
inventario = {"teclado": 10, "ratón": 15, "monitor": 5}

while True:
    print("\n1. Consultar  2. Añadir  3. Retirar  4. Mostrar  5. Salir")
    opcion = input("Elige una opción: ")
    if opcion == "5":
        print("Programa finalizado")
        break
    elif opcion == "4":
        for producto, cantidad in inventario.items():
            print(f"{producto}: {cantidad}")
    elif opcion in ("1", "2", "3"):
        producto = input("Producto: ").lower().strip()
        if producto not in inventario:
            print("Producto no encontrado")
            continue
        if opcion == "1":
            print(f"Existencias: {inventario[producto]}")
            continue
        try:
            cantidad = int(input("Unidades: "))
        except ValueError:
            print("Introduce un número entero")
            continue
        if cantidad < 0:
            print("No se admiten cantidades negativas")
        elif opcion == "2":
            inventario[producto] += cantidad
            print("Unidades añadidas")
        elif cantidad > inventario[producto]:
            print("No hay suficientes unidades")
        else:
            inventario[producto] -= cantidad
            print("Unidades retiradas")
    else:
        print("Opción no válida")
````

## Ejercicio 5

Desarrolla un programa que reciba una dirección IPv4 y determine si tiene un formato válido.

Una dirección IPv4 válida contiene cuatro números enteros entre 0 y 255, separados por puntos.

**Ejemplos**
192.168.1.10 -> Válida
256.10.10.1 -> No válida
192.168.1 -> No válida
192.abc.1.5 -> No válida

---

**Explicación breve:** Dividimos la IP por puntos y comprobamos que haya cuatro partes numéricas entre 0 y 255.

````python
ip = input("Introduce una IPv4: ")
partes = ip.split(".")
valida = len(partes) == 4

# Cada bloque debe contener solo dígitos y estar entre 0 y 255.
if valida:
    for parte in partes:
        if not parte.isdigit() or not 0 <= int(parte) <= 255:
            valida = False
            break

print("IPv4 válida" if valida else "IPv4 no válida")
````

## Ejercicio 6

Un sistema de soporte técnico recibe incidencias con una prioridad numérica entre 1 y 5.

Desarrolla un programa que, a partir de una lista de incidencias, genere un resumen con:
- Número total de incidencias.
- Cantidad de incidencias por prioridad.
- Prioridad más frecuente.
- Porcentaje de incidencias críticas (prioridad 5).

Usa de ejemplo estas incidencias recibidas (los siguientes valores son las prioridades de estas): 1, 3, 5, 2, 5, 3, 4, 5, 1, 3, 5 y 2

---

**Explicación breve:** Contamos las prioridades del 1 al 5, buscamos la más repetida y calculamos el porcentaje de prioridad 5.

````python
incidencias = [1, 3, 5, 2, 5, 3, 4, 5, 1, 3, 5, 2]
conteos = {}
for prioridad in range(1, 6):
    conteos[prioridad] = incidencias.count(prioridad)

mas_frecuente = max(conteos, key=conteos.get)
print("Total de incidencias:", len(incidencias))
for prioridad, cantidad in conteos.items():
    print(f"Prioridad {prioridad}: {cantidad}")
print("Prioridad más frecuente:", mas_frecuente)
print(f"Críticas (prioridad 5): {conteos[5] / len(incidencias) * 100:.2f}%")
````

## Ejercicio 7

Una tienda online aplica descuentos según el importe de la compra:
| Importe | Descuento |
| ------- | :-------: |
| Menos de 50€ | 0 % |
| Desde 50€ hasta menos de 100€ | 5 % |
| Desde 100€ hasta menos de 200€ | 10 % |
| Desde 200€ | 15 % |

Desarrolla un programa que procese una lista de compras y muestre el importe original, el descuento aplicado y el precio final de cada compra.

Al finalizar, debe mostrar el dinero total descontado de todas las compras que se han hecho.

De ejemplo puedes tomar estas compras:
- Primera compra: 25
- Segunda compra: 75
- Tercera compra: 120
- Cuarta compra: 250
- Quinta compra: 90
- Sexta compra: 350

---

**Explicación breve:** Aplicamos `if/elif/else` para decidir el descuento y acumulamos el ahorro total.

````python
compras = [25, 75, 120, 250, 90, 350]
total_descontado = 0
for numero, importe in enumerate(compras, start=1):
    # Elegimos el porcentaje según el tramo de compra.
    if importe < 50:
        porcentaje = 0
    elif importe < 100:
        porcentaje = 5
    elif importe < 200:
        porcentaje = 10
    else:
        porcentaje = 15
    descuento = importe * porcentaje / 100
    final = importe - descuento
    total_descontado += descuento
    print(f"Compra {numero}: original {importe:.2f} €, descuento {descuento:.2f} € ({porcentaje}%), final {final:.2f} €")
print(f"Total descontado: {total_descontado:.2f} €")
````

## Ejercicio 8

Un servidor almacena sus accesos mediante cadenas con el siguiente formato:

IP;MÉTODO;CÓDIGO_HTTP

Desarrolla un programa que procese una lista de registros y muestre:
- Número total de peticiones.
- Cantidad de peticiones GET y POST.
- Número de respuestas con error (códigos HTTP mayor o igual que 400).
- Dirección IP que ha realizado más peticiones.

Trata, por ejemplo, los siguientes registros:
- 192.168.1.10;GET;200
- 192.168.1.20;POST;201
- 192.168.1.10;GET;404
- 192.168.1.30;GET;500
- 192.168.1.10;POST;200
- 192.168.1.20;GET;200

---

**Explicación breve:** Separamos cada registro con `split(";")` y contamos métodos, errores y peticiones por IP.

````python
registros = [
    "192.168.1.10;GET;200", "192.168.1.20;POST;201",
    "192.168.1.10;GET;404", "192.168.1.30;GET;500",
    "192.168.1.10;POST;200", "192.168.1.20;GET;200"
]
metodos = {"GET": 0, "POST": 0}
peticiones_por_ip = {}
errores = 0
for registro in registros:
    ip, metodo, codigo = registro.split(";")
    if metodo in metodos:
        metodos[metodo] += 1
    if int(codigo) >= 400:
        errores += 1
    peticiones_por_ip[ip] = peticiones_por_ip.get(ip, 0) + 1

ip_mas_activa = max(peticiones_por_ip, key=peticiones_por_ip.get)
print("Total de peticiones:", len(registros))
print("GET:", metodos["GET"], "POST:", metodos["POST"])
print("Respuestas con error:", errores)
print("IP con más peticiones:", ip_mas_activa, "-", peticiones_por_ip[ip_mas_activa])
````

## Ejercicio 9

Un sistema de monitorización registra periódicamente la temperatura de una máquina.

Desarrolla un programa que identifique los valores anómalos, considerando anómala cualquier temperatura que se aleje más de 15 grados de la media de todas las mediciones.

El programa debe mostrar la media, los valores anómalos y sus posiciones originales.

Ejemplo de temperaturas: 42, 43, 41, 45, 44, 89, 43, 42, 10 y 44

---

**Explicación breve:** Calculamos la media y usamos `enumerate` para conservar la posición original de cada valor anómalo.

````python
temperaturas = [42, 43, 41, 45, 44, 89, 43, 42, 10, 44]
media = sum(temperaturas) / len(temperaturas)
print(f"Temperatura media: {media:.2f} °C")
print("Valores anómalos (posición original empezando en 0):")
for posicion, temperatura in enumerate(temperaturas):
    # Es anómala si se aleja MÁS de 15 grados de la media.
    if abs(temperatura - media) > 15:
        print(f"Posición {posicion}: {temperatura} °C")
````

## Ejercicio 10

Una plataforma de vídeos tecnologícos registra las categorías de los vídeos que ha visto cada usuario.

Desarrolla un programa que recomiende vídeos de la categoría que más ha consumido el usuario.

No deben recomendarse vídeos que el usuario ya haya visto.

**Ejemplo**
De los siguientes vídeos:
! Nombre del vídeo | Categoría |
| ---------------- | :-------: |
| Python básico | programacion |
| Funciones Python | programacion |
| Docker desde cero | sistemas |
| POO en Python | programacion |
| Listas Python | programacion |
| Redes TCP/IP | sistemas |
| Diccionarios | programacion |
| Linux básico | sistemas |

Puedes suponer que un usuario ha visto "Python básico", "Docker desde cero", "Listas Python" y "Linux básico"

**Explicación breve:** Contamos las categorías de los vídeos vistos y recomendamos los no vistos de la categoría más frecuente.

````python
videos = {
    "Python básico": "programacion",
    "Funciones Python": "programacion",
    "Docker desde cero": "sistemas",
    "POO en Python": "programacion",
    "Listas Python": "programacion",
    "Redes TCP/IP": "sistemas",
    "Diccionarios": "programacion",
    "Linux básico": "sistemas"
}
vistos = ["Python básico", "Docker desde cero", "Listas Python", "Linux básico"]

# Contamos las categorías de los vídeos vistos.
conteos = {}
for titulo in vistos:
    categoria = videos[titulo]
    conteos[categoria] = conteos.get(categoria, 0) + 1

# Si hay empate, elegimos la primera categoría encontrada.
favorita = max(conteos, key=conteos.get)
print("Categoría favorita:", favorita)
print("Vídeos recomendados:")
for titulo, categoria in videos.items():
    if categoria == favorita and titulo not in vistos:
        print("-", titulo)
````
