## Ejercicio 1

Una aplicación recibe nombres de usuario con espacios y diferentes combinaciones de mayúsculas. Desarrolla una función que normalice cada nombre (elimina espacios extremos y pasa a minúsculas), descarte los vacíos y devuelva la lista sin duplicados, conservando el orden de primera aparición.

Ejemplo de entrada: `[' Ana ', 'LUIS', 'ana', '  ', 'Luis', 'Marta']`.


---

**Explicación breve:** Usamos `strip()` y `lower()` para normalizar; una lista evita duplicados sin perder el orden.

```python
def normalizar_nombres(nombres):
    resultado = []
    for nombre in nombres:
        nombre = nombre.strip().lower()  # Quita espacios extremos y pasa a minúsculas
        if nombre != "" and nombre not in resultado:
            resultado.append(nombre)  # Conserva la primera aparición
    return resultado

usuarios = [' Ana ', 'LUIS', 'ana', '  ', 'Luis', 'Marta']
print(normalizar_nombres(usuarios))
```

## Ejercicio 2

Un comercio recibe líneas de pedido como diccionarios con "producto", "precio" y "unidades". Calcula el importe total de cada línea y del pedido completo. Las líneas con precio negativo o unidades no positivas deben rechazarse con un mensaje, sin interrumpir el resto del cálculo.

Prueba por ejemplo para estos datos:
{'producto': 'SSD', 'precio': 69.90, 'unidades': 2},
{'producto': 'RAM', 'precio': 45, 'unidades': 3},
{'producto': 'Cable', 'precio': -4, 'unidades': 1}


---

**Explicación breve:** Recorremos cada línea, rechazamos las cantidades no válidas y acumulamos los importes válidos.

```python
pedido = [
    {'producto': 'SSD', 'precio': 69.90, 'unidades': 2},
    {'producto': 'RAM', 'precio': 45, 'unidades': 3},
    {'producto': 'Cable', 'precio': -4, 'unidades': 1}
]
total = 0
for linea in pedido:
    if linea['precio'] < 0 or linea['unidades'] <= 0:
        print(f"Línea rechazada: {linea['producto']} (precio o unidades no válidos)")
        continue  # Seguimos con la siguiente línea
    importe = linea['precio'] * linea['unidades']
    total += importe
    print(f"{linea['producto']}: {importe:.2f} €")
print(f"Total del pedido: {total:.2f} €")
```

## Ejercicio 3

Un sistema almacena lecturas "(sensor, temperatura)" de varios dispositivos. Calcula para cada sensor su temperatura media y muestra solo los sensores cuya media supere los 50 grados, ordenados de mayor a menor media.


---

**Explicación breve:** Agrupamos las temperaturas por sensor, calculamos medias y ordenamos con `sorted()`.

```python
# Datos de ejemplo: (sensor, temperatura)
lecturas = [('A', 48), ('B', 65), ('A', 56), ('C', 42), ('B', 59), ('C', 46)]
temperaturas = {}
for sensor, temperatura in lecturas:
    if sensor not in temperaturas:
        temperaturas[sensor] = []
    temperaturas[sensor].append(temperatura)

medias = {}
for sensor, valores in temperaturas.items():
    media = sum(valores) / len(valores)
    if media > 50:
        medias[sensor] = media

# Ordenamos por la media de mayor a menor
for sensor, media in sorted(medias.items(), key=lambda dato: dato[1], reverse=True):
    print(f"Sensor {sensor}: {media:.2f} °C")
```

## Ejercicio 4

Un formulario recibe edades introducidas como texto. Crea una función que convierta cada valor a entero y acepte solamente edades entre 0 y 120. Debe devolver "None" si la entrada es inválida.

Prueba la función con "['25', ' 41 ', '-3', 'abc', '121']".


---

**Explicación breve:** Convertimos con `int()` dentro de `try/except` y devolvemos `None` fuera del rango.

```python
def validar_edad(texto):
    try:
        edad = int(texto.strip())  # Convertimos el texto a entero
        if 0 <= edad <= 120:
            return edad
    except (ValueError, AttributeError):
        pass  # Texto no convertible a entero
    return None

edades = ['25', ' 41 ', '-3', 'abc', '121']
for valor in edades:
    print(f"{valor!r} -> {validar_edad(valor)}")
```

## Ejercicio 6

Un servicio recibe una lista de direcciones de correo. Agrúpalas por dominio y cuenta cuántas pertenecen a cada uno. Ignora direcciones sin una única "@" o sin texto a ambos lados de ella; la comparación de dominios no distingue mayúsculas.


---

**Explicación breve:** Comprobamos el formato básico del correo, extraemos el dominio y acumulamos sus apariciones.

```python
# Correos de ejemplo
correos = ['ana@Gmail.com', 'luis@gmail.com', 'marta@empresa.es',
           'sin_arroba', '@empresa.es', 'a@@empresa.es', 'pepe@EMPRESA.ES']
dominios = {}
for correo in correos:
    if correo.count('@') != 1:
        continue
    usuario, dominio = correo.split('@')
    if usuario == '' or dominio == '':
        continue  # Debe haber texto a ambos lados
    dominio = dominio.lower()
    dominios[dominio] = dominios.get(dominio, 0) + 1

for dominio, cantidad in sorted(dominios.items()):
    print(f"{dominio}: {cantidad}")
```

## Ejercicio 7

Dos almacenes guardan sus existencias en diccionarios "{producto: unidades}". Genera un inventario global que sume los productos comunes y conserve los exclusivos de cada almacén. Muestra el resultado ordenado alfabéticamente.

Usa por ejemplo:
- norte = {'raton': 8, 'teclado': 5, 'monitor': 2}
- sur = {'raton': 4, 'webcam': 7, 'monitor': 3}


---

**Explicación breve:** Copiamos el primer diccionario y sumamos las existencias del segundo con `get()`.

```python
norte = {'raton': 8, 'teclado': 5, 'monitor': 2}
sur = {'raton': 4, 'webcam': 7, 'monitor': 3}
global_inventario = norte.copy()  # Partimos de las existencias del norte
for producto, unidades in sur.items():
    global_inventario[producto] = global_inventario.get(producto, 0) + unidades

for producto in sorted(global_inventario):
    print(f"{producto}: {global_inventario[producto]}")
```

## Ejercicio 8

Un archivo de texto llamado "accesos.txt" contiene una dirección IP por línea. Crea un programa que cuente las apariciones de cada IP, ignore las líneas vacías y escriba en "resumen.txt" las IP y sus cantidades.


---

**Explicación breve:** Leemos el archivo línea a línea, ignoramos vacíos y escribimos el resumen en otro archivo.

```python
# Creamos un archivo de ejemplo para poder probar el ejercicio
with open('accesos.txt', 'w', encoding='utf-8') as fichero:
    fichero.write('192.168.1.10\n192.168.1.20\n\n192.168.1.10\n10.0.0.1\n')

conteos = {}
with open('accesos.txt', 'r', encoding='utf-8') as fichero:
    for linea in fichero:
        ip = linea.strip()  # Eliminamos el salto de línea
        if ip != '':
            conteos[ip] = conteos.get(ip, 0) + 1

with open('resumen.txt', 'w', encoding='utf-8') as fichero:
    for ip, cantidad in sorted(conteos.items()):
        fichero.write(f"{ip}: {cantidad}\n")

print('Resumen guardado en resumen.txt')
for ip, cantidad in sorted(conteos.items()):
    print(f"{ip}: {cantidad}")
```

## Ejercicio 9

Una empresa recibe un CSV "ventas.csv" con columnas "producto", "unidades" y "precio". 

Calcula la facturación acumulada de cada producto y muestra el producto con mayor facturación.


---

**Explicación breve:** Leemos el CSV con `DictReader`, acumulamos unidades por precio y buscamos el máximo.

```python
import csv

# Creamos un CSV de prueba (sustituible por el archivo real)
ventas_ejemplo = [
    ['producto', 'unidades', 'precio'],
    ['teclado', 2, 30], ['raton', 5, 15], ['teclado', 3, 30], ['monitor', 1, 120]
]
with open('ventas.csv', 'w', newline='', encoding='utf-8') as fichero:
    csv.writer(fichero).writerows(ventas_ejemplo)

facturacion = {}
with open('ventas.csv', 'r', newline='', encoding='utf-8') as fichero:
    for fila in csv.DictReader(fichero):
        producto = fila['producto']
        importe = int(fila['unidades']) * float(fila['precio'])
        facturacion[producto] = facturacion.get(producto, 0) + importe

for producto, total in sorted(facturacion.items()):
    print(f"{producto}: {total:.2f} €")
if facturacion:
    ganador = max(facturacion, key=facturacion.get)
    print(f"Mayor facturación: {ganador} ({facturacion[ganador]:.2f} €)")
```

## Ejercicio 10

Una agenda guarda contactos en un archivo "contactos.json" como lista de objetos con "nombre" y "telefono".

Desarrolla dos funciones: una para añadir un contacto (sin permitir nombres duplicados ignorando mayúsculas) y otra para buscarlo por nombre.

Prueba ambas funciones.

**Explicación breve:** Guardamos los contactos en JSON y comparamos nombres sin distinguir mayúsculas.

```python
import json

# Creamos la agenda si todavía no existe
try:
    with open('contactos.json', 'r', encoding='utf-8') as fichero:
        json.load(fichero)
except FileNotFoundError:
    with open('contactos.json', 'w', encoding='utf-8') as fichero:
        json.dump([], fichero, indent=4, ensure_ascii=False)

def buscar_contacto(nombre):
    with open('contactos.json', 'r', encoding='utf-8') as fichero:
        contactos = json.load(fichero)
    for contacto in contactos:
        if contacto['nombre'].lower() == nombre.strip().lower():
            return contacto
    return None

def anadir_contacto(nombre, telefono):
    nombre = nombre.strip()
    if nombre == '' or buscar_contacto(nombre) is not None:
        print(f"No se puede añadir '{nombre}': vacío o duplicado")
        return False
    with open('contactos.json', 'r', encoding='utf-8') as fichero:
        contactos = json.load(fichero)
    contactos.append({'nombre': nombre, 'telefono': telefono})
    with open('contactos.json', 'w', encoding='utf-8') as fichero:
        json.dump(contactos, fichero, indent=4, ensure_ascii=False)
    print(f"Contacto añadido: {nombre}")
    return True

# Pruebas: se conserva el archivo entre ejecuciones
anadir_contacto('Ana', '600111222')
anadir_contacto('ana', '600999999')  # Duplicado
print('Búsqueda:', buscar_contacto('ANA'))
```
