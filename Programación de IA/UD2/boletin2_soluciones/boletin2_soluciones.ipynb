## Normalización de nombres

**Archivo:** `ejercicio1.py`

```python
def normalizar_usuarios(nombres):
    resultado = []
    vistos = set()

    for nombre in nombres:
        normalizado = nombre.strip().lower()
        if normalizado and normalizado not in vistos:
            resultado.append(normalizado)
            vistos.add(normalizado)

    return resultado


if __name__ == "__main__":
    usuarios = [' Ana ', 'LUIS', 'ana', '  ', 'Luis', 'Marta']
    print(normalizar_usuarios(usuarios))
```

## Cálculo de pedidos

**Archivo:** `ejercicio2.py`

```python
def calcular_pedido(lineas):
    total_pedido = 0

    for linea in lineas:
        producto = linea["producto"]
        precio = linea["precio"]
        unidades = linea["unidades"]

        if precio < 0 or unidades <= 0:
            print(f"Línea rechazada ({producto}): precio o unidades no válidos.")
            continue

        importe = precio * unidades
        total_pedido += importe
        print(f"{producto}: {importe:.2f} €")

    print(f"Total del pedido: {total_pedido:.2f} €")
    return total_pedido


if __name__ == "__main__":
    pedido = [
        {"producto": "SSD", "precio": 69.90, "unidades": 2},
        {"producto": "RAM", "precio": 45, "unidades": 3},
        {"producto": "Cable", "precio": -4, "unidades": 1},
    ]
    calcular_pedido(pedido)
```

## Media de temperaturas por sensor

**Archivo:** `ejercicio3.py`

```python
def medias_superiores(lecturas):
    temperaturas = {}

    for sensor, temperatura in lecturas:
        temperaturas.setdefault(sensor, []).append(temperatura)

    medias = {
        sensor: sum(valores) / len(valores)
        for sensor, valores in temperaturas.items()
    }

    return sorted(
        ((sensor, media) for sensor, media in medias.items() if media > 50),
        key=lambda elemento: elemento[1],
        reverse=True,
    )


if __name__ == "__main__":
    lecturas = [
        ("S1", 48), ("S1", 54), ("S2", 62),
        ("S2", 66), ("S3", 45), ("S3", 51),
    ]
    for sensor, media in medias_superiores(lecturas):
        print(f"{sensor}: {media:.2f} °C")
```

## Validación de edades

**Archivo:** `ejercicio4.py`

```python
def validar_edad(valor):
    try:
        edad = int(valor.strip())
    except (ValueError, AttributeError):
        return None

    if 0 <= edad <= 120:
        return edad
    return None


if __name__ == "__main__":
    valores = ["25", " 41 ", "-3", "abc", "121"]
    print([validar_edad(valor) for valor in valores])
```

## Informe de puntuaciones

**Archivo:** `ejercicio5.py`

```python
def informe_puntuaciones(votos):
    if any(not isinstance(voto, int) or not 1 <= voto <= 5 for voto in votos):
        raise ValueError("Todas las puntuaciones deben ser enteros entre 1 y 5.")

    frecuencias = {puntuacion: votos.count(puntuacion) for puntuacion in range(1, 6)}
    media = sum(votos) / len(votos) if votos else None

    if votos:
        frecuencia_maxima = max(frecuencias.values())
        mas_frecuentes = [
            puntuacion
            for puntuacion, cantidad in frecuencias.items()
            if cantidad == frecuencia_maxima
        ]
    else:
        mas_frecuentes = []

    return frecuencias, media, mas_frecuentes


if __name__ == "__main__":
    votos = [5, 4, 5, 3, 4, 2, 5, 4]
    frecuencias, media, mas_frecuentes = informe_puntuaciones(votos)
    print("Votos:", frecuencias)
    print("Media:", media)
    print("Puntuación/es más frecuente/s:", mas_frecuentes)
```

## Agrupación de correos por dominio

**Archivo:** `ejercicio6.py`

```python
def contar_dominios(correos):
    dominios = {}

    for correo in correos:
        if correo.count("@") != 1:
            continue

        usuario, dominio = correo.split("@")
        usuario = usuario.strip()
        dominio = dominio.strip().lower()

        if not usuario or not dominio:
            continue

        dominios[dominio] = dominios.get(dominio, 0) + 1

    return dominios


if __name__ == "__main__":
    correos = [
        "ana@gmail.com", "luis@outlook.com", "marta@GMAIL.com",
        "invalido", "@dominio.com", "usuario@", "otro@gmail.com",
    ]
    print(contar_dominios(correos))
```

## Inventario conjunto

**Archivo:** `ejercicio7.py`

```python
def inventario_global(norte, sur):
    inventario = norte.copy()

    for producto, unidades in sur.items():
        inventario[producto] = inventario.get(producto, 0) + unidades

    return dict(sorted(inventario.items()))


if __name__ == "__main__":
    norte = {"raton": 8, "teclado": 5, "monitor": 2}
    sur = {"raton": 4, "webcam": 7, "monitor": 3}
    print(inventario_global(norte, sur))
```

## Recuento de accesos desde archivo

**Archivo:** `ejercicio8.py`

```python
from collections import Counter


def resumir_accesos(archivo_entrada="accesos.txt", archivo_salida="resumen.txt"):
    with open(archivo_entrada, "r", encoding="utf-8") as archivo:
        ips = [linea.strip() for linea in archivo if linea.strip()]

    recuento = Counter(ips)

    with open(archivo_salida, "w", encoding="utf-8") as archivo:
        for ip, cantidad in sorted(recuento.items()):
            archivo.write(f"{ip}: {cantidad}\\n")

    return recuento


if __name__ == "__main__":
    resultado = resumir_accesos()
    print("Resumen guardado en resumen.txt")
    print(dict(resultado))
```

## Facturación por producto desde CSV

**Archivo:** `ejercicio9.py`

```python
import csv
from collections import defaultdict


def facturacion_por_producto(nombre_archivo="ventas.csv"):
    facturacion = defaultdict(float)

    with open(nombre_archivo, "r", newline="", encoding="utf-8") as archivo:
        lector = csv.DictReader(archivo)
        for fila in lector:
            producto = fila["producto"]
            unidades = int(fila["unidades"])
            precio = float(fila["precio"])
            facturacion[producto] += unidades * precio

    return dict(facturacion)


if __name__ == "__main__":
    facturacion = facturacion_por_producto()
    for producto, total in sorted(facturacion.items()):
        print(f"{producto}: {total:.2f} €")

    if facturacion:
        producto_maximo = max(facturacion, key=facturacion.get)
        print(f"Mayor facturación: {producto_maximo} ({facturacion[producto_maximo]:.2f} €)")
```

## Agenda de contactos JSON

**Archivo:** `ejercicio10.py`

```python
import json
from pathlib import Path


def cargar_contactos(archivo="contactos.json"):
    ruta = Path(archivo)
    if not ruta.exists():
        return []

    with ruta.open("r", encoding="utf-8") as fichero:
        return json.load(fichero)


def anadir_contacto(nombre, telefono, archivo="contactos.json"):
    contactos = cargar_contactos(archivo)

    if any(contacto["nombre"].casefold() == nombre.casefold() for contacto in contactos):
        return False

    contactos.append({"nombre": nombre, "telefono": telefono})

    with open(archivo, "w", encoding="utf-8") as fichero:
        json.dump(contactos, fichero, ensure_ascii=False, indent=4)

    return True


def buscar_contacto(nombre, archivo="contactos.json"):
    contactos = cargar_contactos(archivo)

    for contacto in contactos:
        if contacto["nombre"].casefold() == nombre.casefold():
            return contacto

    return None


if __name__ == "__main__":
    archivo_prueba = "contactos.json"
    print("Añadido:", anadir_contacto("Ana", "600123123", archivo_prueba))
    print("Duplicado:", anadir_contacto("ANA", "600999999", archivo_prueba))
    print("Búsqueda:", buscar_contacto("ana", archivo_prueba))
```
