```python
'''
# # Actividad seleccionando Series y DataFrames
# Se tiene la siguiente información de sensores, temperaturas y humedad:
# {
# "sensor": [“sensor1", “sensor2", “sensor3", “sensor4", “sensor5"],
# "temperatura": [45.5, 52.3, 48.7, 61.2, 39.8],
# "humedad": [60, 55, 62, 58, 65]
# }
# Devuelve:
# ▪ El número total de sensores.
# ▪ Una Series con las temperaturas.
# ▪ La temperatura media.
# ▪ Añadir una columna que valga True cuando la temperatura sea superior a
# 50 y False cuando valga 50 o menos.
'''
frame_datos = pd.DataFrame([
    {
        "Sensor": 1,
        "temperatura": 45.5,
        "humedad": 60
    },
        {
        "Sensor": 2,
        "temperatura": 52.3,
        "humedad": 55
    },
        {
        "Sensor": 3,
        "temperatura": 48.7,
        "humedad": 62
    },
        {
        "Sensor": 4,
        "temperatura": 61.2,
        "humedad": 58
    }
])

print(frame_datos)
print("////////////////")
print(f"El numero total de sensores es:  {frame_datos["Sensor"].size}")
print("////////////////")
print(frame_datos["temperatura"])
print("////////////////")
temperaturas = frame_datos["temperatura"]
print(f"La media de las temperaturas es de: {temperaturas.mean()}")

```

```python
# Usando sqlite3 de Python, crea una base de datos sensores.db que tenga una
# tabla sensores con los datos de sensor, temperatura y humedad usados en
# otros ejercicios, inserta los valores, haz commit y finalmente realiza un SELECT
# usando pandas.read_sql_query() cuando la temperatura sea mayor de 60.
# Muestra el resultado y cierra la conexión.
import sqlite3

conexion = sqlite3.connect("Sensores.db")
cursor = conexion.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS sensores (
        sensor TEXT,
        temperatura REAL,
        humedad REAL
    )
""")

datos = [
    ("Sensor 1", 60, 42),
    ("Sensor 2", 78, 41),
    ("Sensor 3", 66, 39),
]

cursor.executemany("""
    INSERT INTO sensores (sensor, humedad, temperatura)
    VALUES (?, ?, ?)
""", datos)

conexion.commit()
conexion.close()
```
