Una empresa tecnológica utiliza cinco servidores para ejecutar diferentes aplicaciones de IA.

Se ha registrado el porcentaje de memoria RAM utilizada por cada servidor:

| Servidor | RAM utilizada (%) |
| :------: | :---------------: |
| servidor1	| 45 |
| servidor2	| 78 |
| servidor3	| 92 |
| servidor4	| 63 |
| servidor5	| 85 |

Realiza:

- Crea una Series con los porcentajes de RAM, utilizando los nombres de los servidores como índices.
- Muestra todos los valores.
- Muestra el porcentaje de RAM del servidor3 mediante loc.
- Muestra el porcentaje del segundo servidor mediante iloc.
- Calcula la media de utilización de RAM.
- Muestra el valor máximo y el mínimo.
- Muestra el número total de servidores.


```python
import pandas as pd

# 1. Crear una Series con los porcentajes de RAM
ram = pd.Series(
    [45, 78, 92, 63, 85],
    index=["servidor1", "servidor2", "servidor3", "servidor4", "servidor5"]
)

# 2. Mostrar todos los valores
print("Porcentaje de RAM de los servidores:")
print(ram)

# 3. Mostrar la RAM del servidor3 mediante loc
print("\nRAM del servidor3:")
print(ram.loc["servidor3"])

# 4. Mostrar la RAM del segundo servidor mediante iloc
print("\nRAM del segundo servidor:")
print(ram.iloc[1])

# 5. Calcular la media de utilización
print("\nMedia de utilización de RAM:")
print(ram.mean())

# 6. Mostrar el valor máximo y mínimo
print("\nRAM máxima:")
print(ram.max())

print("\nRAM mínima:")
print(ram.min())

# 7. Mostrar el número total de servidores
print("\nNúmero total de servidores:")
print(len(ram))
```
