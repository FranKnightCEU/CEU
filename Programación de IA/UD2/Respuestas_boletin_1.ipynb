```python
# Ejercicio 1, contraseña


input_usr = input("Introduzca su contraseña: ")

segura = False
mayuscula = False
minuscula = False 
numero = False 
caracteresp = False 

for i in input_usr:
    if i.isupper():
        mayuscula = True
    if i.islower():
        minuscula = True 
    if i.isdigit():
        numero = True 
    if i.isalnum():
        caracteresp = False
    else: 
         caracteresp = True


while segura == False:
    if len(input_usr) < 8:
        print("La contraseña es demasiado corta")
        
    if mayuscula == False:
        print("La contraseña debe incluir una letra en mayúscula")
        
    if minuscula == False:
            print("La contraseña debe incluir una letra en minúscula")
            
    if numero == False:
            print("La contraseña debe incluir un número")
            
    if caracteresp == False: 
         print("La contraseña debe incluir al menos un caracter especial")
         
    else: 
         print("Contraseña aceptada")
         segura = True 
         break
    input_usr = input("Introduzca su contraseña: ")
    
    
```

```python
tiempos = [120, 340, 560, 230, 890, 120, 450, 670, 210, 180]

tiempos_media = 0
suma = 0
for i in tiempos:
    suma += i

tiempos_media = suma / len(tiempos)
Maxim = max(tiempos)
Minim = min(tiempos)
print(f"La media de tiempo es {tiempos_media}")
print(f"El tiempo mas alto ha sido {Maxim}")
print(f"El tiempo mas bajo ha sido {Minim}")

for j in tiempos:
    if j > 500:
        print(f"{j} es velocidad aceptable")
    else:
        print(f"{j} no es una velocidad aceptable")
```
