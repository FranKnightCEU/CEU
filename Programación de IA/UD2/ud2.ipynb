```python
datos = [0.95,0.72,0.91,0.84,0.98,0.67]

umbral = 85

datos_porcentaje = []
for i in datos:
    porcentaje = i*100
    datos_porcentaje.append(porcentaje)
print(datos_porcentaje)
confianzas_aceptadas = []
for j in datos_porcentaje:
    if j >= umbral:
        confianzas_aceptadas.append(j) 
print(f"Confianzas aceptadas: {confianzas_aceptadas}")



for posicion,val in enumerate(datos_porcentaje,start=1) :
    print(f"predicción {posicion}\n\t: {"aceptada" if val > umbral  else "rechazada"}")
```

funciones:
- Calcular la media de una serie de confianzas que se le pasen
- Cuente cuantas predicciones aceptadas y rechazadas
- Evalue las predicciones pasandole la confianza y el umbral por parametro



```python

umbral_muestra = int(input("Introduzca un umbral de confianza: "))
confianzas_muestra = []

confianza = 1
while confianza != 0:
    confianza = int(input("Introduzca un valor de confianza: "))
    if confianza != 0:
        confianzas_muestra.append(confianza)


def evaluar_predicción(confianzas, umbral):
    for i in confianzas:
        if i <= umbral:
            print(f"La confianza de valor {i} no supera el umbral con valor de {umbral}")
        else:
            print(f"La confianza de valor {i} supera el umbral con valor de {umbral}")


def calcular_media(valores_media_calc):
    num_valores = len(valores_media_calc)

    if num_valores == 0:
        print("No hay valores para calcular la media.")
        return

    conjunto = 0
    for j in valores_media_calc:
        conjunto += j

    media = conjunto / num_valores
    print(f"La media es {media}")


evaluar_predicción(confianzas_muestra, umbral_muestra)
calcular_media(confianzas_muestra)


```

# Ejercicio práctico
# Empleando el módulo json queremos guardar información de IPS obtenidas de un log a un fichero mis_accesos.json. Vamos a guardar dos atributos, "ip" y "ubicacion". Guardaremos 3 registros.
# Una vez almacenado, quiero abrir ese fichero "mis_accesos.json" y leer todos los datos
 

```python
import json 

datos_log = [
    {"ip": "214.2.23.35", "pais":"españa"},
    {"ip": "204.1.33.90", "pais":"españa"},
    {"ip": "226.1.1.11", "pais":"españa"},
]

with open("mis_accesos.json", "w", encoding="utf-8") as fichero:
    json.dump(datos_log, fichero, indent=4, ensure_ascii=False)

with open("mis_accesos.json", "r", encoding="utf-8") as fichero:
    datos_fichero = json.load(fichero)
    for fila in datos_fichero:
        print(fila)
```
