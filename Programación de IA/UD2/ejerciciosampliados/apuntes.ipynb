```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# LEer los datos
datos = pd.read_csv('lecturas_maquinas.csv', sep=';')

#limpiar los datos 
# duplicates() isna() 


print(f"Duplicados: {datos.duplicated().sum()}")
print(f"Valores nulos: {datos.isna().sum()}")
# limpar los datos malos
datos_depurados = datos.drop_duplicates().dropna()

# registros eliminados: {len(datos) - len(datos_depurados)}")
```

```python
# datos por aula 
datos_aula_a = datos_depurados[datos_depurados['Aula'] == 'A']
datos_co2_a = datos_aula_a['co2_ppm']
print(f"Media de CO2 en Aula A: {round(datos_co2_a.mean(), 2)} ppm")

# incidencias >1000 alta y >1200 critica
# Aula a

print(f"Incidencias en el aula a: {datos_aula_a[(datos_aula_a['co2_ppm'] > 1000) & (datos_aula_a['co2_ppm'] <= 1200)].shape[0]} incidencias altas")
print(f"Incidencias en el aula a: {datos_aula_a[datos_aula_a['co2_ppm'] > 1200].shape[0]} incidencias críticas")


import matplotlib.pyplot as plt

#crear un gráfico de lineas con los datos de co2 en el aula a
plt.plot(datos_aula_a['dia'], datos_aula_a['co2_ppm'])
```
