```python
import numpy as np

mi_array = np.array([10, 20, 30, 40, 50, 60])

# Devuelve los tres primeros valores
tres = mi_array[:3]

# Muestra los cuatro últimos valores
cuatro = mi_array[-4:]

# Muestra los cuatro últimos valores, empezando desde el último
cuatro_invertidos = mi_array[:-5:-1]

# Muestra un valor sí y otro no empezando por el segundo
alternos = mi_array[1::2]

# Devuelve los valores desde la posición cuarta hasta el final
desde_cuarta = mi_array[3:]

print("Tres primeros:", tres)
print("Cuatro últimos:", cuatro)
print("Cuatro últimos invertidos:", cuatro_invertidos)
print("Valores alternos:", alternos)
print("Desde la cuarta posición:", desde_cuarta)
```

```python
import numpy as np

array = np.array([41, 22, 89, 99, 22])

# 1. Valores menores de 50 y además pares
print(array[(array < 50) & (array % 2 == 0)])

# 2. Valores que terminen en 1 o empiecen por 8
print(array[(array % 10 == 1) | ((array >= 80) & (array < 90))])

# 3. Valores que no sean pares (impares)
print(array[array % 2 != 0])
```

```python
'''Una aplicación de inteligencia artificial analiza imágenes para identificar objetos. Después de procesar varias imágenes, devuelve las siguientes puntuaciones de confianza:

[0.95, 0.82, 0.91, 0.65, 0.98, 0.73]

Desarrolla un programa utilizando que:
- Muestre el número total de predicciones realizadas.
- Calcule la confianza media, máxima y mínima.
- Muestre todas las confianzas expresadas en porcentaje.
- Muestre las confianzas superiores o iguales a 0.90.
- Indique cuántas predicciones tienen una confianza inferior a 0.80.
'''

import numpy as np

valores = np.array([0.95, 0.82, 0.91, 0.65, 0.98, 0.73])
print(len(valores))

media = valores.mean()
maxima = valores.max()
minima = valores.min()

print(f"maxima : {maxima}")
print(f"minima : {minima}")
print(f"media : {media}")

porcentaje = valores * 100
for i in porcentaje: 
    print(f"pocentual : {i} %")

print(valores[valores >= 0.90])

menor = valores[valores <= 0.80]
print(len(menor))
```
