## Ejercicio práctico

```python
# Ejercicio práctico
# A partir del ejercicio anterior de comprensiones de listas y diccionarios, realiza
# tres funciones que:
# ▪ Evalúen las predicciones pasándole la confianza y el umbral por
# parámetro
# ▪ Calcule la media de una serie de confianzas que se le pase
# ▪ Cuente las predicciones aceptadas y rechazadas
total = {
    "aceptadas":0,
    "rechazadas":0
}

def evaluar_prediccion(confianza,umbral):
    if confianza >= UMBRAL :
        return "aceptada"
    else :
        return "rechazada"

def media_confianzas(predicciones):
    return sum(predicciones)/len(predicciones)

def contar_predicciones_segun_confianza(predicciones):
    for i in predicciones :
        if evaluar_prediccion(i,UMBRAL) == "aceptada" :
            total["aceptadas"] += 1
        else :
            total["rechazadas"] += 1
    return total["aceptadas"],total["rechazadas"]

for i in puntuaciones:
    print(f"La predicción {i} se {evaluar_prediccion(i,UMBRAL)}")

print(media_confianzas(puntuaciones))

aceptadas, rechazadas = contar_predicciones_segun_confianza(puntuaciones)
print(f"Aceptadas: {aceptadas}")
print(f"Rechazadas: {rechazadas}")
```
