## 1. Introducción y Contexto de Negocio
 En ese cuaderno, enfocaremos nuestra atención exclusivamente en los **algoritmos de clasificación**, aquellos que utilizamos como cuando el resultado que esperamos predecir es una etiqueta discreta dentro de un conjunto finito de categorías.

Esta práctica está diseñada para dar cumplimiento directo al Resultado de Aprendizaje 2 (RA 2) del módulo, poniéndoos a prueba en la selección de herramientas de Machine Learning.

Hoy abordaremos el mismo problema de clasificación de negocio utilizando tres paradigmas computacionales radicalmente distintos basándonos en el manual de la asignatura:

*   **k-Nearest Neighbors (k-NN):** Un modelo basado en instancias que clasifica buscando las observaciones más cercanas.

Memoriza los datos y, al recibir un elemento nuevo, calcula matemáticamente la distancia de este frente a todos los elementos históricos

Consume mucha capacidad de procesamiento y memoria al requerir cargar todo el conjunto de entrenamiento en memoria para realizar cada clasificación

  *Imagina que te mudas a un barrio nuevo y quieres adivinar a qué partido político votarás. El algoritmo k-NN ($k=3$) simplemente buscaría a las 3 personas que viven físicamente más cerca de tu casa; si 2 de ellas votan al Partido A y 1 vota al Partido B, el modelo predecirá que tú votarás al Partido A por "mayoría de vecindad"*

  


*   **Árboles de Decisión (CART):** Classification and Regression Trees. Un modelo "caja blanca" basado en reglas explícitas, ideal para la explicabilidad al cliente.

La estructura se divide en nodos (variables de entrada que plantean opciones de decisión), ramas (los valores que pueden tomar dichas variables) y hojas (el resultado final o valor de salida).

La particularidad fundamental del algoritmo CART frente a otros árboles es que genera particiones estrictamente binarias; es decir, cada nodo siempre se divide en exactamente dos ramas.

Es un modelo de "caja blanca", lo que significa que el camino lógico desde la entrada hasta la predicción final es fácilmente interpretable y auditable por un cliente o equipo de negocio.

*El triaje de urgencias de un hospital. Llega un paciente y el árbol evalúa nodos binarios: "¿Fiebre > 39ºC?" (Sí/No). Si la rama es "Sí", pasa al siguiente nodo: "¿Dificultad respiratoria?" (Sí/No). Si la rama es "Sí", se llega a la hoja final: "Ingreso inmediato".*

*   **Clasificador Naïve Bayes:** Un modelo estadístico probabilístico que logra una altísima velocidad de cálculo computacional.

Se le denomina "ingenuo" (naïve) porque asume una hipótesis de independencia absoluta entre los atributos o características del conjunto de datos.

Su principal ventaja computacional es que solo requiere un conjunto de datos de entrenamiento pequeño para estimar las medias y las varianzas necesarias.

Al basarse en la probabilidad de que el mejor modelo es simplemente el más probable es un algoritmo extremadamente rápido y simple, utilizado con mucha frecuencia en la clasificación de texto.

*Un filtro antispam de correo electrónico. El algoritmo evalúa un email analizando las palabras por separado. Si la palabra "Viagra" tiene un 90% de probabilidad de ser spam y la palabra "Príncipe" tiene un 85%, el modelo multiplica estas probabilidades ignorando si gramaticalmente la frase "Príncipe Viagra" tiene algún sentido en el mundo real.*

## 2. Definición del Problema y Preparación de Datos
💡 **Analogía:** Entrenar un modelo sin dividir los datos en Entrenamiento (Train) y Prueba (Test) es como daros a vosotros los alumnos las respuestas del examen final el día antes. Sacaréis un 10, pero habréis *memorizado*, no *aprendido*. Por eso, siempre guardaremos un porcentaje de datos "bajo llave" para evaluar el modelo al final.

**Ejemplo Guiado: Dataset Iris**
En nuestro ejemplo guiado, utilizaremos el clásico dataset Iris referenciado en el libro:
* **Tarea ($T$):** Clasificar especies de flores (Setosa, Versicolor, Virginica).
* **Experiencia ($E$):** 150 muestras con 4 características.
* **Rendimiento ($R$):** Precisión global ($Accuracy$) del modelo.

```python
# ==========================================
# EJEMPLO GUIADO: Partición del dataset Iris
# ==========================================

# Importamos la función específica para cargar nuestro dataset de prueba (Iris)
from sklearn.datasets import load_iris
# Importamos la herramienta clave para trocear y separar nuestros datos
from sklearn.model_selection import train_test_split

# 1. Carga y extracción de datos
iris = load_iris()

# X (Mayúscula): Matriz de características. Son los datos que usamos para predecir (longitud y ancho de pétalos/sépalos).
# 💡 Analogía: Estos son los enunciados de los problemas.
X_iris = iris.data

# y (minúscula): Vector objetivo. Es la respuesta correcta (la especie real de la flor).
# 💡 Analogía: Estas son las soluciones a los problemas.
y_iris = iris.target

# 2. Partición: 80% entrenamiento (Train) y 20% test (Test)
# ¡IMPORTANTE! Aquí NO hay IA, NO hay entrenamiento y NO calculamos el Accuracy.
# Simplemente estamos separando los "apuntes para estudiar" de "las preguntas del examen final".
X_train_iris, X_test_iris, y_train_iris, y_test_iris = train_test_split(
    X_iris,          # Le pasamos la matriz de características (las preguntas).
    y_iris,          # Le pasamos el vector objetivo (las soluciones).
    test_size=0.20,  # Reservamos el 20% de los datos bajo llave. El algoritmo JAMÁS los verá al estudiar.
    random_state=123 # Fijamos una semilla (123) para que la partición aleatoria sea siempre idéntica y el experimento sea reproducible.
)

# 3. Verificación de la partición
# Comprobamos cuántas muestras han ido a parar al bloque de estudio y cuántas al examen final.
print(f"Material de estudio (Train): {X_train_iris.shape[0]} muestras. Preguntas de examen (Test): {X_test_iris.shape[0]} muestras.")
```

Lo único que hace este bloque es la partición de los datos (separar la Experiencia $E$ de entrenamiento de los datos de prueba).
Llamar a train_test_split es como cuando yo, el profesor, separo los apuntes con los que vais a estudiar en casa (X_train) de las preguntas que os pondrá en el examen final (X_test). En este paso todavía no estamos haciendo el examen ni calculando vuestra nota (el Accuracy), solo estamos preparando los materiales".

### 🕵️‍♂️ Análisis de Estructuras de Datos: ¿Qué hay dentro de "iris"?
En Machine Learning, nunca operamos datos a ciegas. Si en la celda anterior pasas el cursor del ratón por encima de la variable `iris`, verás que Colab te muestra una radiografía de lo que acabas de descargar de Scikit-Learn:

> `Bunch: iris (8 items) sklearn.utils._bunch.Bunch{'data': ndarray with shape (150, 4), 'target': ndarray with shape (150,), ...}`

**¿Cómo leemos esta información como ingenieros de datos?**
* **El objeto `Bunch`:** Un objeto Bunch en Scikit-Learn es una subclase del diccionario estándar de Python (dict). La ventaja estructural del Bunch es que expone sus claves como atributos de objeto, permitiendo acceder a los valores usando la notación de punto (iris.data) en lugar de la notación estricta de corchetes de los diccionarios tradicionales (iris['data']).
* **`'data': ndarray with shape (150, 4)`:** Aquí está nuestra matriz de características ($X$). `ndarray` nos dice que es una matriz matemática de NumPy. `shape (150, 4)` nos chiva sus dimensiones exactas: **150 muestras (filas)** de flores analizadas y **4 características (columnas)** matemáticas medidas por cada flor.
* **`'target': ndarray with shape (150,)`:** Aquí están las soluciones ($y$). Un vector unidimensional con 150 respuestas correctas (la especie de cada flor).

💡 **Buena práctica:** Acostúmbrate a inspeccionar el `shape` (la forma) de tus matrices de datos. El 90% de los errores al programar redes neuronales o algoritmos de Machine Learning vienen por intentar operar matrices con dimensiones (shapes) incompatibles.

# Bloque 02. Algoritmo k-Nearest Neighbors (k-NN)
💡 **Analogía:** Imagina que te mudas a un barrio nuevo y no sabes si es una zona de estudiantes o residencial familiar. ¿Qué haces? Miras a tus $k$ vecinos más cercanos. Si 3 son estudiantes y 1 es familia, asumes que es zona de estudiantes. Así de simple (y "perezoso") es k-NN.

El método clasifica un nuevo elemento asignándolo al grupo mayoritario entre sus $k$ vecinos más cercanos mediante el cálculo de distancias. Es muy sensible al parámetro $k$ y requiere todo el conjunto de entrenamiento en memoria para clasificar nuevos datos.

**Evaluación: Matriz de Confusión y Accuracy**
* **Matriz de Confusión:** Tabla que cruza predicciones vs realidad. La diagonal principal son los aciertos.
* **Accuracy (Precisión Global):** Porcentaje total de aciertos.

```python
# ==========================================
# EJEMPLO GUIADO: k-NN en Iris
# ==========================================

# Importa el algoritmo k-NN, un modelo basado en instancias que asume que la clase depende de los k vecinos más próximos
from sklearn.neighbors import KNeighborsClassifier

# Importa las funciones que actuarán como "correctores del examen" para medir el Rendimiento (R).
from sklearn.metrics import accuracy_score, confusion_matrix

# 1. INSTANCIAR EL MODELO
# Creamos el algoritmo definiendo su hiperparámetro crítico: buscará las 5 flores más cercanas para clasificar.
knn_iris = KNeighborsClassifier(n_neighbors=5)

# 2. FASE DE ENTRENAMIENTO
# Le pasamos los datos de estudio (Experiencia E). Al ser un algoritmo "perezoso", guarda los datos en memoria para calcular distancias luego.
knn_iris.fit(X_train_iris, y_train_iris)

# 3. FASE DE PREDICCIÓN (INFERENCIA)
# Ponemos a prueba el modelo pidiéndole que clasifique las flores del conjunto de test que nunca ha visto.
y_pred_knn_iris = knn_iris.predict(X_test_iris)

# 4. FASE DE EVALUACIÓN
# Cruzamos las respuestas reales con las predicciones. Muestra una tabla detallando los aciertos (diagonal) y los errores.
print("Matriz de Confusión:\n", confusion_matrix(y_test_iris, y_pred_knn_iris))

# Calcula la "nota final": el porcentaje global de predicciones correctas sobre el total de casos evaluados.
print("Accuracy:", accuracy_score(y_test_iris, y_pred_knn_iris))
```

🎯 Precisión vs. Exhaustividad (Recall)
Precisión: De todo lo que el modelo predijo como positivo, ¿cuántos lo eran realmente? (Mide la fiabilidad de la predicción).

Exhaustividad (Recall): De todos los positivos reales que existían en los datos, ¿cuántos logró atrapar el modelo? (Mide la capacidad de búsqueda).

Prioriza la EXHAUSTIVIDAD (Recall) cuando el coste de un Falso Negativo es catastrófico.

Prioriza la PRECISIÓN cuando el coste de un Falso Positivo es inaceptable.

### 💡 La gran pregunta de negocio: ¿Por qué no buscamos el 100% de Exhaustividad siempre?

Podrías pensar: *"Configuremos el modelo para que lo detecte absolutamente todo"*. Pero en Machine Learning nada es gratis: existe una cuerda tensa entre **Precisión** y **Exhaustividad (Recall)**.

* **La paradoja del detector alarmista:** Si quieres un 100% de Exhaustividad detectando fraude bancario, basta con programar un modelo que bloquee el 100% de las tarjetas del banco. Habrás atrapado a todos los ladrones (Exhaustividad del 100%), pero habrás bloqueado a millones de clientes inocentes (Precisión cercana al 0%).
* **¿Qué error es más caro?** En la empresa, cada casilla de la Matriz de Confusión tiene un precio en euros:
  * **Cuando el Falso Negativo es más caro (Ej. Medicina, Ciberseguridad):** Preferimos pagar el coste de revisar falsas alarmas (Falsos Positivos) antes que dejar pasar un solo caso letal. **Optimizamos Exhaustividad (Recall).**
  * **Cuando el Falso Positivo es más caro (Ej. Filtro de Spam, Justicia):** Preferimos que se nos escape algún caso antes que acusar o bloquear en falso algo crítico. **Optimizamos Precisión.**

# Bloque 03. Árboles de Decisión (CART)
💡 **Analogía:** Los árboles de decisión son como jugar al clásico juego de mesa "¿Quién es quién?". El algoritmo hace preguntas binarias sucesivas (¿El nivel de alcohol es mayor a 13? ¿El color es intenso?) para ir descartando opciones hasta quedarse con una predicción clara.

A diferencia de k-NN, los árboles generan reglas explícitas fáciles de interpretar por el negocio (caja blanca).
📚 **Documentación:** [DecisionTreeClassifier](https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeClassifier.html)

```python
# Importamos las herramientas necesarias
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Cargamos el conjunto de datos Iris
iris = load_iris()

# Separamos las características (X) y las etiquetas (y)
X_iris = iris.data
y_iris = iris.target

# Dividimos los datos en entrenamiento (75%) y prueba (25%)
X_train_iris, X_test_iris, y_train_iris, y_test_iris = train_test_split(X_iris, y_iris, test_size=0.25, random_state=42)

# ///////////////////////////////////////////////////////////7
# =====================================================================
# EJEMPLO GUIADO: Árbol de Decisión (CART) en Iris
# =====================================================================

# Importamos la clase del algoritmo de Árboles de Decisión (CART) y la función para trazar y dibujar el árbol.
from sklearn.tree import DecisionTreeClassifier, plot_tree

# Importamos la librería de Python para la generación de gráficos y visualizaciones.
import matplotlib.pyplot as plt

# 1. INSTANCIAR EL MODELO
# Creamos el objeto del árbol de decisión pasando por parámetros 'random_state=123' para asegurar la reproducibilidad
# y 'max_depth=3' para limitar la profundidad máxima de las ramas y evitar el sobreajuste (overfitting).
cart_iris = DecisionTreeClassifier(random_state=123, max_depth=3)

# 2. FASE DE ENTRENAMIENTO
# Ajustamos el modelo pasando por parámetros los datos de estudio: la matriz de características (X_train_iris)
# y el vector de etiquetas reales (y_train_iris), para que el algoritmo aprenda las reglas de partición jerárquica.
cart_iris.fit(X_train_iris, y_train_iris)

# 3. FASE DE PREDICCIÓN (INFERENCIA)
# Ponemos a prueba el modelo entrenado pasando por parámetros el conjunto de test (X_test_iris)
# que el algoritmo jamás ha visto, guardando las predicciones de especies en la variable 'y_pred_cart_iris'.
y_pred_cart_iris = cart_iris.predict(X_test_iris)

# 4. REPRESENTACIÓN VISUAL DEL ÁRBOL
# Creamos una figura en blanco indicando sus dimensiones de ancho y alto en pulgadas pasando por parámetros 'figsize=(8, 4)'.
plt.figure(figsize=(8, 4))

# Dibujamos la estructura jerárquica del árbol pasando por parámetros el modelo entrenado ('cart_iris'),
# activando 'filled=True' para colorear los nodos según la clase mayoritaria, e indicando los nombres
# de las características predictivas ('feature_names=iris.feature_names') y las clases objetivo ('class_names=iris.target_names').
plot_tree(cart_iris, filled=True, feature_names=iris.feature_names, class_names=iris.target_names)

# Añadimos un título descriptivo a la gráfica pasando por parámetros el texto del título.
plt.title("Árbol de Decisión - Iris")

# Renderizamos y mostramos la ventana gráfica resultante en pantalla.
plt.show()
```

# Píldora: ¿Qué es Scikit-Learn y por qué es el estándar de la industria?

Scikit-learn estandarizó la forma en que los programadores interactúan con los algoritmos. Sin importar la complejidad matemática del modelo, el flujo de trabajo en código siempre consta de los mismos tres pasos lógicos que hoy has practicado:

1. **Instanciar el modelo:** `modelo = AlgoritmoClassifier()`
2. **Entrenar (Ajustar a los datos):** `modelo.fit(X_train, y_train)`
3. **Inferir (Predecir y Evaluar):** `predicciones = modelo.predict(X_test)`

```python
#@title SOLUCIÓN RETO 01

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

# =====================================================================
# PASO 1: Carga y Extracción de Datos
# =====================================================================
# 1.1 Invoca la función que hemos importado arriba para cargar los datos.
#     Guarda el resultado en la variable 'wine'.
#     Recuerda que esto descargará un "expediente estructurado" (Bunch).
wine = load_wine()

# 1.2 Extrae la matriz de características matemáticas (los datos químicos).
#     Pista: Busca y extrae el atributo '.data' dentro de tu objeto 'wine'.
X_wine = wine.data

# 1.3 Extrae el vector objetivo (las etiquetas reales de la clase de vino).
#     Pista: Busca y extrae el atributo '.target' dentro de tu objeto 'wine'.
y_wine = wine.target

# =====================================================================
# PASO 2: Partición del Dataset (Train / Test Split)
# =====================================================================
# Ahora vamos a separar los "apuntes de estudio" de las "preguntas del examen".
# Llama a la función 'train_test_split' pasándole 4 ingredientes clave:
#   1. Tus características (X_wine)
#   2. Tus etiquetas (y_wine)
#   3. El tamaño del conjunto de prueba (busca el parámetro 'test_size' y ponlo al 0.25)
#   4. La semilla de aleatoriedad (busca el parámetro 'random_state' y asígnale 42)
#
# Cuidado con el orden de las 4 variables que devuelve la función, debe ser exacto:
X_train_wine, X_test_wine, y_train_wine, y_test_wine = train_test_split(
    X_wine, y_wine, test_size=0.25, random_state=42
)

# =====================================================================
# PASO 3: Verificación de Dimensiones
# =====================================================================
# Como ingenieros de datos, nunca avanzamos sin comprobar las dimensiones (shape).
# Utiliza la función 'print()' combinada con el atributo '.shape[0]' para mostrar
# por pantalla cuántas filas (muestras) han ido a parar a 'X_train_wine' y cuántas a 'X_test_wine'.

print(f"Material de estudio (Train): {X_train_wine.shape[0]} muestras.")
print(f"Preguntas de examen (Test): {X_test_wine.shape[0]} muestras.")
```

```python
#@title SOLUCIÓN RETO 02.1
# Importamos la clase del algoritmo k-NN (k vecinos más próximos) desde el módulo 'neighbors' de scikit-learn.
from sklearn.neighbors import KNeighborsClassifier

# =====================================================================
# PASO 1: Instanciar el modelo (Configurar el "molde" matemático)
# =====================================================================
# El método k-NN es muy sensible al parámetro 'k', que indica el número de vecinos seleccionados.
# Pista: Llama a la clase 'KNeighborsClassifier()' que acabamos de importar arriba y pásale
# por parámetros 'n_neighbors=3' para que evalúe los 3 vinos más cercanos.
knn_wine_3 = KNeighborsClassifier(n_neighbors=3)

# =====================================================================
# PASO 2: Entrenar el modelo (Fase de Aprendizaje con la Experiencia E)
# =====================================================================
# En scikit-learn, todos los modelos aprenden invocando el método '.fit()' sobre el objeto creado.
# Como k-NN no precisa construir un modelo global, en este paso guardará todo el conjunto de entrenamiento en memoria.
# Pista: Escribe el nombre de tu modelo ('knn_wine_3') seguido de '.fit()' y pásale por parámetros,
# tus "apuntes de estudio": las características (X_train_wine) y las respuestas reales (y_train_wine).
knn_wine_3.fit(X_train_wine, y_train_wine)

# Mensaje de control por consola para confirmar que la celda se ha ejecutado sin errores de sintaxis.
print("✅ Modelo k-NN entrenado.")
```

```python
#@title SOLUCIÓN RETO 02.2
# Importamos la función que actuará como "corrector del examen" para medir el Rendimiento (R).
from sklearn.metrics import accuracy_score

# =====================================================================
# PASO 1: Realizar Predicciones (Fase de Inferencia / El Examen)
# =====================================================================
# Invocamos el método '.predict()' de nuestro modelo entrenado ('knn_wine_3')
# pasándole únicamente los datos de entrada reservados para la evaluación ('X_test_wine').
predicciones_knn = knn_wine_3.predict(X_test_wine)

# =====================================================================
# PASO 2: Evaluar el Rendimiento (Calcular la Nota Final)
# =====================================================================
# Calculamos la exactitud (accuracy) comparando las etiquetas reales ('y_test_wine')
# con las predicciones generadas por el algoritmo ('predicciones_knn').
precision_knn = accuracy_score(y_test_wine, predicciones_knn)

# Imprimimos por pantalla el porcentaje de acierto (de 0.0 a 1.0).
print(f"🎯 Accuracy de k-NN (k=3): {precision_knn}")

# =====================================================================
# 🧪 RETO DE EXPERIMENTACIÓN (Ajuste de Hiperparámetros):
# =====================================================================
# El algoritmo k-NN es extremadamente sensible al parámetro 'k'.
# Vuelve a la celda anterior (Paso 2.1), cambia 'n_neighbors=1' (o 'n_neighbors=7'),
# ejecuta aquella celda y luego vuelve a ejecutar esta. ¿Sube o baja la nota del examen?
```
