```python
import json

print("Bienvenido a mi test de Python")

usuarios = []

while True:
    nombre = input("Introduce tu nombre (o 'salir' para terminar): ")

    if nombre.lower() == "salir":
        break

    try:
        edad = int(input("Introduce tu edad: "))
    except ValueError:
        print("Por favor, introduce una edad válida.")
        continue

    try:
        comandante = input("Introduce tu comandante: ")
    except ValueError:
        print("Por favor, introduce un nombre válido.")
        continue

    datos = {
        "nombre": nombre,
        "edad": edad,
        "comandante" : comandante
    }

    usuarios.append(datos)

    with open("datos.json", "w", encoding="utf-8") as archivo:
        json.dump(usuarios, archivo, ensure_ascii=False, indent=4)

    print("Usuario guardado correctamente.")

print("Todos los datos han sido guardados.")
```
